
# Lab 8 – Hidden Markov Model (HMM)
Simulation of phoneme transitions for the word 'speech'


```python
import numpy as np
import pandas as pd
```

```python

# Define HMM parameters
states = ['/s/', '/p/', '/ie:/', '/tS/']
observations = ['Energy', 'Pitch', 'Duration']
init_prob = np.array([1, 0, 0, 0])

trans_probs = np.array([
    [0.1, 0.8, 0.1, 0.0],  
    [0.0, 0.0, 0.8, 0.2],  
    [0.0, 0.0, 0.2, 0.8],  
    [0.2, 0.0, 0.0, 0.8]   
])

emiss_probs = np.array([
    [0.7, 0.2, 0.1],
    [0.5, 0.3, 0.2],
    [0.3, 0.5, 0.2],
    [0.4, 0.4, 0.2]
])
```

```python

# Display HMM matrices
print('Initial Probabilities:')
display(pd.DataFrame([init_prob], columns=states))
print('Transition Probabilities:')
display(pd.DataFrame(trans_probs, columns=states, index=states))
print('Emission Probabilities:')
display(pd.DataFrame(emiss_probs, columns=observations, index=states))
```

```python

# Generate a single phoneme and observation sequence
def generate_sequence():
    seq = []
    obs_seq = []
    curr_state = 0
    for step in range(4):
        seq.append(states[curr_state])
        emis = np.random.choice(observations, p=emiss_probs[curr_state])
        obs_seq.append(emis)
        if step < 3:
            curr_state = np.random.choice(len(states), p=trans_probs[curr_state])
    return seq, obs_seq

seq, obs_seq = generate_sequence()
print('Generated Phoneme Sequence:', seq)
print('Generated Observation Sequence:', obs_seq)
```

Inference: The HMM simulates the progression of speech's phonemes, capturing probabilistic transitions and emitting acoustic observations. The structure matches typical speech sequence flows. Generated sequences help illustrate temporal dependencies and statistical properties in speech processing.
